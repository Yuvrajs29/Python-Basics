```python
##Accessing the elements in the list 
fruits = ["Apple","Banana","Cherry","Kiwi","Guava"]
```

```python
print(fruits[0])
print(fruits[1])
print(fruits[3])
print(fruits[-1])
```

```python
print(fruits[1:]) 
print(fruits[1:3])  ##from start till 3 index(exclusive)
print(fruits[:4])
print(fruits[-1:-5:-1]) ## [start(inclusive) :stop(exclusive) : step(direction)]
print(fruits[:3])
print(fruits[1::-1])
print(fruits[1:4:-1])
```

```python
fruits[1]="watermelon"
print(fruits)
```

```python
fruits[1:]="watermelon"
print(fruits)
```

```python
fruits[1:]=["watermelon"]
print(fruits)
```

```python
fruits = ["Apple","Banana","Cherry","Kiwi","Guava"]
```

```python
## Append : it is used to add element to the end of list
fruits.append("orange")
print(fruits)
```

```python
# insert : to insert it on any position
fruits.insert(1,"watermelon")
print(fruits)
```

```python
fruits.remove("watermelon")  ## it removes only the first occurence of item
print(fruits)
```

```python
## Pop = it removes and returns the element at given index , if index is not specified then it will remove the last item
fruits.pop(2)
print(fruits)
```

```python
fruits.index("Banana")
fruits.insert(1,"Banana")
print(fruits)
```

```python
count = fruits.count("Banana")
print(count)
```

```python
fruits.sort()
print(fruits)
```

```python
fruits.reverse()
print(fruits)
```

```python
fruits.clear()
print(fruits)
```
