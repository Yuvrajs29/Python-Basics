```python
## While LOOP 

## the while loop continue to execute as long as the condition is true .

count=0
while count<5:
    print(count)
    count=count+1
```

```python
count=0 
while count%2==0 :
    print(count)
    count=count+1
```

```python
## Loop Control Statementes
## break
## the break statement exist the loop prematurely 
## break statement 
for i in range(10):
    if i==5:
        break 
    print(i)
  
```

```python
## continue 

## continue statement skips the current iteration and continue with the next.

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
## pass 
## the pass statement is a null operation it does nothing.

for i in range(5):
    if i==3:
        print("the number is :",i)
        pass
    print(i)    
```

```python
## Nested Loops 
## a loop inside a loop 

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
###Examples - Calculate the sum of first n natural numbers using a while and for loop
## while loop
n=5
sum=0
count=1
while count<=n:
    sum=sum+count
    count=count+1

print("Sum of n natural number :",sum)    
```

```python
## for loop
n=5
sum=0
for i in range(1,n+1):
    sum=sum+i
print(sum)    
```

```python
## Example - Prime number between 1 and 100
for i in range(2,101):
    for j in range(2,i):
            if i%j==0:
                break
    else:
            print(i)
```
