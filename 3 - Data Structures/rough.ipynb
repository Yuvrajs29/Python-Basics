```python
def temp_conv(temp,unit):
    if unit=='C':
        return temp*9/5 + 32
    elif unit=='F':
        return (temp-32)*5/9
    else:
        return None
```

```python
print(temp_conv(25,'C'))
print(temp_conv(77,'F'))
```
